```python
a = qrisp.QuantumModulus(17)
b = qrisp.QuantumModulus(17)

a.m = 5
b.m = 1

qrisp.h(a)
qrisp.h(b)

temp = b.duplicate()
qrisp.cx(b, temp)


print(b)
b+=a
print(b)

mes_res = qrisp.multi_measurement([temp, a, b])

for temp, a, b in mes_res.keys():
    
    if temp is np.nan or a is np.nan:
        continue
    if (temp + a)%17 != b:
        raise
```
